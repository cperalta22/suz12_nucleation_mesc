```r
pdf("/hdd/Itzel/mRNA-422336955/featurecounts/VolcanosJul2025/bp_alternative_mockVrescue.pdf", width = 8, height = 8)

ggplot(bpmvr, aes(x= log2FoldChange, y = -log(padj), color = Significative)) + 
geom_point() +
scale_color_manual(values = c("FALSE" = "olivedrab3", "TRUE" = "olivedrab")) + 
annotate("rect", 
           xmin = -Inf, 
           xmax = -1.5, 
           ymin = -log10(0.005), 
           ymax = Inf,
           fill = "blue", 
           alpha = 0.05) + 
annotate("rect", 
           xmin = 1.5, 
           xmax = Inf, 
           ymin = -log10(0.005), 
           ymax = Inf,
           fill = "red", 
           alpha = 0.05)+
theme_classic()

dev.off()
```

```r
mvr <- mvr |> mutate(class = case_when(
    (isbp == T & directionality == 'UP') ~ 'bpup',
    (isbp == T & directionality == 'DOWN') ~ 'bpdown',
    (isbp == T & directionality == 'NONE') ~ 'bp',
    (isbp == F & directionality == 'UP') ~ 'bulkup',
    (isbp == F & directionality == 'DOWN') ~ 'bulkdown',
    .default = "bulk"))
table(mvr$class)
table(mvr$isbp)
table(mvr$directionality)
```
