```r
library(corrplot)
library(RColorBrewer)
```

```r
intersection_matrix <- as.matrix(read.table("/hdd/Itzel/NuevoTodo/chromhmm_bienPortadosDeseq_44/jul25_chromhmm/Intervene_results/Intervene_pairwise_frac_matrix.txt"))
```

```r
intersection_matrix
```

```r
corrplot(intersection_matrix, method ="color", title="Pairwise intersection-Fraction of overlap", tl.col="black", tl.cex=0.8, is.corr = FALSE, diag=FALSE, addrect=1, mar=c(0,0,2,1), rect.col = "black")
```

```r
corrplot(intersection_matrix,
         method ="color", 
         title="Pairwise intersection-Fraction of overlap", 
         tl.col="black", tl.cex=0.8, is.corr = FALSE, 
         diag=FALSE,
         addrect=1,
         mar=c(0,0,2,1),
         col = colorRampPalette(c("Cyan", "Darkviolet"))(200),
         rect.col = "black")
```
